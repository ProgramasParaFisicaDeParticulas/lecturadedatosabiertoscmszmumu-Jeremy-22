```python
#panda nos permite manipular el archivo csv https://pandas.pydata.org
import pandas as pd
#numpy nos permite hacer operaciones usando vectores https://numpy.org
import numpy as np
#plotly nos permite crear graficas interactivas https://plotly.com/python/
import plotly.express as px
%matplotlib inline
#usamos la liga que está directamente en la página.
particles = pd.read_csv('https://opendata.cern.ch/record/5208/files/Zmumu.csv',delimiter=',')
particles.head()
```

```python
fig_5=px.histogram(particles['pt1'],particles['pt1'])
m_in_category = pd.cut(particles['pt1'], bins=[0, 150], labels=['Momento transversal'])
fig_5=px.histogram(particles['pt1'],particles['pt1'], color=m_in_category, labels={'x': 'Momento transversal', 'color':''})
fig_5.update_layout(
    xaxis_title="$P_{T}[GeV]$",
    yaxis_title="Frecuencia"
)
fig_5.show()
```

```python
#Se establecen los bins convenientes para limpiar nuestros datos.
phi = (-np.pi, np.pi)
eta = (-2.1, 2.1)
pt = (35, 90)
fil_particles = particles[(particles['eta1'].between(*eta)) 
                          & (particles['eta2'].between(*eta))
                          & (particles['phi1'].between(*phi))
                          & (particles['pt1'].between(*pt))
                          & (particles['pt2'].between(*pt))
                          & (particles['phi2'].between(*phi))
                          & ((particles['Q1'] * particles['Q2']) == -1)]

fil_particles.head()

```

```python
fig_5=px.histogram(fil_particles['pt1'],fil_particles['pt1'])

fig_5.show()
```

```python
ev = len(fil_particles)
print(ev)
```

```python
combined_df = pd.DataFrame({
    'Transverse Momentum (pt)': pd.concat([fil_particles['pt1'], fil_particles['pt2']]),
    'Tipo': ['pt1'] * len(fil_particles) + ['pt2'] * len(fil_particles)
})


fig = px.histogram(combined_df, x='Transverse Momentum (pt)', color='Tipo', labels={'x': 'Momentp transversal', 'color':''}, title="Comparación de pt1 y pt2", barmode='overlay')
fig.update_xaxes(range=[33, 80])
fig.update_layout(
    xaxis_title='$P_T$ [GeV]',
    yaxis_title='Frecuencia'
)
fig.show()
```

```python
combined_df = pd.DataFrame({
    'Carga (Q)': pd.concat([fil_particles['Q1'], fil_particles['Q2']], ignore_index=True),
    'Tipo': ['Q1'] * len(fil_particles) + ['Q2'] * len(fil_particles)
})

fig = px.histogram(combined_df, x='Carga (Q)', color='Tipo', title="Distribución de las Cargas de los Muones", barmode='overlay')

fig.update_xaxes(range=[-3, 3])
fig.update_yaxes(range=[0, 3000])


fig.update_layout(
    xaxis_title='Carga (Q)',
    yaxis_title='Frecuencia'
)
moda_q1 = particles['Q1'].mode()[0]
moda_q2 = particles['Q2'].mode()[0]

print(f'Moda de Q1: {moda_q1}')
print(f'Moda de Q2: {moda_q2}')
fig.show()
```

```python
def m_inv(pt1,pt2,eta1,eta2,phi1,phi2):
    return np.sqrt(2*pt1*pt2*(np.cosh(eta1-eta2)-np.cos(phi1-phi2)))
m_in = m_inv(fil_particles['pt1'],fil_particles['pt2'],fil_particles['eta1'],fil_particles['eta2'],fil_particles['phi1'],fil_particles['phi2'])
bins = [0, 150]
labels = ['masa invariante']
m_in_category = pd.cut(m_in, bins=bins, labels=labels)
fig_5=px.histogram(m_in, color=m_in_category, labels={'x': 'Masa invariante', 'color':''})
fig_5.update_layout(
    xaxis_title="$M_{T}[GeV]$",
    yaxis_title="Frecuencia"
)
fig_5.show()
```

```python
md = pd.DataFrame(m_in)

# Calcular estadísticas
median_values = md.median()
std_values = md.std()
var_values = md.var()
mean_values = md.mean()  # Promedio


# Imprimir resultados
print("Mediana:", median_values)
print("\nDesviación estándar:\n", std_values)
print("\nVarianza:\n", var_values)
print("\nPromedio:\n", mean_values)
```
